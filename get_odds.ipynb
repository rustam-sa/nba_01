```python
import requests
import os
from dotenv import load_dotenv

class OddsBlazeAPI:
    def __init__(self):
        # Load the API key from the .env file
        load_dotenv("odds_blaze.env")  # Specify the path to your .env file
        self.api_key = os.getenv("API_KEY")
        if not self.api_key:
            raise ValueError("API_KEY not found in the .env file")
        
        self.base_url = "https://sgp.oddsblaze.com/v1/"

    def fetch_sgp_prices(self, sportsbook, sgp_fields=None, price_format="american", odds_body=None):
        """
        Fetch same game parlay prices from the OddsBlaze API.

        :param sportsbook: The name of the sportsbook (e.g., 'Hard Rock Bet').
        :param sgp_fields: A list of SGP fields (optional if using POST body).
        :param price_format: Format of the price ('american' or 'decimal').
        :param odds_body: List of SGP odds for POST body (optional).
        :return: JSON response from the API.
        """
        params = {
            "key": self.api_key,
            "sportsbook": sportsbook,
            "price": price_format,
        }
        
        if sgp_fields:
            params["sgp"] = ",".join(sgp_fields)

        headers = {
            "Content-Type": "application/json"
        }
        
        # Send request
        try:
            if odds_body:
                response = requests.post(self.base_url, params=params, json={"odds": odds_body}, headers=headers)
            else:
                response = requests.get(self.base_url, params=params, headers=headers)
            
            response.raise_for_status()
            return response.json()
        
        except requests.exceptions.RequestException as e:
            print(f"Error fetching data: {e}")
            return None
```
