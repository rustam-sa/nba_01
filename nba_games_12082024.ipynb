```python
from datetime import datetime
from data_manager import DataManager

dm = DataManager()
```

```python
subject_teams = [
"NYK", "TOR"
]
```

```python
exam_players = dm.get_players_in_teams(subject_teams)
player_logs = dm.get_player_logs(exam_players)
```

```python
import pandas as pd

game_data_players = []

for player in exam_players:
    player_id = dm.get_player_id(player)
    if player_id:
        player_game_data = dm.get_and_save_player_data(player_id)
        
        # Check if the result is a DataFrame and if it's not empty
        if isinstance(player_game_data, pd.DataFrame):
            if not player_game_data.empty:
                game_data_players.append(player_game_data)
            else:
                print(f"Player {player} ({player_id}) returned an empty DataFrame.")
        else:
            print(f"Player {player} ({player_id}) did not return a DataFrame.")
```

```python
print(len(game_data_players))
```

```python
teams = dm.pull_teams()
```

```python
selected_teams = []
for team in teams:
    if team['abbreviation'] in subject_teams:
        selected_teams.append(team['nickname'])
```

```python
game_data_teams = []

for team in selected_teams:
    team_id = dm.get_team_id(team)
    team_game_data = dm.get_and_save_team_data(team_id)
    
    # Check if the result is a DataFrame and if it's not empty
    if isinstance(team_game_data, pd.DataFrame):
        if not team_game_data.empty:
            game_data_teams.append(team_game_data)
        else:
            print(f"Team {team} ({team_id}) returned an empty DataFrame.")
    else:
        print(f"Team {team} ({team_id}) did not return a DataFrame.")
```

```python
game_data = dict(zip(exam_players, game_data_players))
```

```python
print(len(game_data))
```

```python
game_data = dict(zip(exam_players, game_data_players))
for i, team in enumerate(selected_teams):
    game_data[team] = game_data_teams[i]
```
